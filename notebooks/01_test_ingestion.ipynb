```python
from youtube_transcript_api import YouTubeTranscriptApi
from urllib.parse import urlparse, parse_qs

def extract_video_id(url):
    query = urlparse(url)
    if query.hostname == 'youtu.be':
        return query.path[1:]
    if query.hostname in ('www.youtube.com', 'youtube.com'):
        if query.path == '/watch':
            return parse_qs(query.query)['v'][0]
        if query.path.startswith('/embed/'):
            return query.path.split('/')[2]
    return None

try:
    video_url = "https://www.youtube.com/watch?v=TaWs8tIrnoA"
    video_id = extract_video_id(video_url)

    transcript = YouTubeTranscriptApi().fetch(video_id)  
    text = " ".join([snippet.text for snippet in transcript])
    print("Transcript fetched successfully!")
    print(text[:500])

except Exception as e:
    print("YouTube ingestion failed:", e)
```

```python
from langchain_core.documents import Document
import whisper

try:
    audio_path = r"C:\Users\BIT\Downloads\Charlie Puth - How Long [Official Video] - Charlie Puth.mp3"
    model = whisper.load_model("base")
    result = model.transcribe(audio_path)
    text = result.get("text", "")

    audio_docs = [Document(page_content=text, metadata={"source": audio_path})]

    print(f"Audio ingestion successful — {len(audio_docs)} document loaded with {len(text)} characters")
    print("Sample text:\n", audio_docs[0].page_content[:500])

except Exception as e:
    print("Audio ingestion failed:", e)
```
